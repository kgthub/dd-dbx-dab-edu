```python
query = """
    select *, (Quantity * UnitPrice) as TotalPrice from 
    training.kgt.online_retail_bronze
    where InvoiceNo not like 'C%'
"""

df = spark.sql(query)
```

```python
%skip
from pyspark.sql.functions import col

df = spark.table("training.kgt.online_retail_bronze") \
    .filter(~col("InvoiceNo").like("C%")) \
    .withColumn("TotalPrice", col("Quantity") * col("UnitPrice"))
```
